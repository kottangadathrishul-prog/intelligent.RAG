```python
import os
import re
import torch

from google.colab import files

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
    pipeline
)

from langchain_huggingface import HuggingFacePipeline

print("All libraries loaded successfully.")
```

```python
print("Upload your PDF:")

uploaded = files.upload()

pdf_file = list(uploaded.keys())[0]

print("\nUploaded:", pdf_file)
```

```python
loader = PyPDFLoader(pdf_file)

documents = loader.load()

print("Pages loaded:", len(documents))


def clean_text(text):
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'Page\s+\d+', '', text, flags=re.IGNORECASE)
    return text.strip()


for doc in documents:
    doc.page_content = clean_text(doc.page_content)

print("\nPDF successfully cleaned.")

print("\nPreview:")
print(documents[0].page_content[:1500])
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=100,
    separators=[
        "\n\n",
        "\n",
        ". ",
        "? ",
        "! ",
        " ",
        ""
    ]
)

chunks = text_splitter.split_documents(documents)

print("Total chunks:", len(chunks))

print("\nExample chunk:")
print(chunks[0].page_content)
```

```python
EMBEDDING_MODEL = "sentence-transformers/all-MiniLM-L6-v2"

embeddings = HuggingFaceEmbeddings(
    model_name=EMBEDDING_MODEL,
    model_kwargs={
        "device": "cuda" if torch.cuda.is_available() else "cpu"
    },
    encode_kwargs={
        "normalize_embeddings": True
    }
)

print("Embedding model loaded.")
```

```python
vectorstore = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)

print("FAISS vector database created.")

print("Stored chunks:", len(chunks))
```

```python

# Cell 8 - Detailed Retrieval

retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={
        "k": 4
    }
)

print("✅ Detailed retriever created!")
```

```python
LLM_MODEL = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(
    LLM_MODEL
)

model = AutoModelForSeq2SeqLM.from_pretrained(
    LLM_MODEL
)

print("LLM loaded successfully.")
```

```python
# Cell 10 - Detailed FLAN-T5 Generation

def generate_answer(prompt_text):

    inputs = tokenizer(
        prompt_text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = model.generate(
            **inputs,

            # Generate a longer answer
            max_new_tokens=300,

            # Minimum answer length
            min_new_tokens=40,

            # Better quality
            num_beams=5,

            # Deterministic generation
            do_sample=False,

            # Reduce repeated sentences
            repetition_penalty=1.15,

            # Encourage complete answers
            length_penalty=1.2,

            early_stopping=True
        )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer.strip()


print("✅ Detailed FLAN-T5 generation ready!")
```

```python
# Cell 11 - Detailed RAG Prompt

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
"""You are an intelligent educational PDF question-answering assistant.

Your job is to provide a clear, detailed, and useful answer to the
user's question using ONLY the information provided in the CONTEXT.

Follow these rules carefully:

1. First understand what the user is asking.
2. Answer the question directly.
3. Give a detailed explanation instead of a one-line answer.
4. Normally provide 2 to 4 paragraphs.
5. Include important definitions, concepts, examples, steps,
   advantages, disadvantages, or applications when they are
   present in the CONTEXT.
6. If the question asks "what is", start with a clear definition.
7. If the question asks "explain", provide a detailed explanation.
8. If the question asks "how", provide the steps in order.
9. If the question asks for differences, explain each side clearly.
10. If an example is present in the CONTEXT, include it.
11. Do not invent information.
12. Do not use information outside the provided document.
13. If the answer cannot be found in the CONTEXT, say:
    "The answer is not available in the provided document."
14. Make the answer easy for a student to understand.
15. Do not repeat the same sentence unnecessarily.

CONTEXT:
-------------------------
{context}
-------------------------

QUESTION:
{question}

DETAILED ANSWER:
"""
)

print("✅ Detailed RAG prompt created!")
```

```python
# Cell 12 - Detailed RAG Pipeline

def format_docs(docs):

    formatted = []

    for doc in docs:

        page = doc.metadata.get("page", "Unknown")

        formatted.append(
            f"[Page {page + 1}]\n{doc.page_content}"
        )

    return "\n\n".join(formatted)


def answer_question(question):

    # Retrieve relevant PDF content
    docs = retriever.invoke(question)

    # Combine retrieved information
    context = format_docs(docs)

    # Create detailed prompt
    final_prompt = prompt.format(
        context=context,
        question=question
    )

    # Generate detailed answer
    answer = generate_answer(final_prompt)

    return answer, docs


print("✅ Detailed RAG pipeline ready!")
```

```python
# Cell 13 - Detailed PDF RAG Chatbot

print("=" * 65)
print("🤖 INTELLIGENT PDF RAG ASSISTANT")
print("=" * 65)

print("\nAsk detailed questions about your PDF.")
print("Type 'exit' to stop.\n")


while True:

    question = input("You: ").strip()

    if question.lower() == "exit":

        print("\n👋 Assistant stopped.")
        break

    if not question:

        print("⚠️ Please enter a question.\n")
        continue

    try:

        answer, source_docs = answer_question(question)

        print("\n🤖 Assistant:")
        print("-" * 65)
        print(answer)
        print("-" * 65)

        # Show source pages
        pages = []

        for doc in source_docs:

            page_number = doc.metadata.get("page", None)

            if page_number is not None:
                pages.append(page_number + 1)

        pages = sorted(set(pages))

        print("\n📄 Sources:")
        print("Pages:", pages)

        print("\n")

    except Exception as e:

        print("\n❌ Error:")
        print(e)

        print("\nPlease try again.\n")
```
